# Self-Attention Matrix Dimensions
L = 50 tokens, d = 64, $d_k$ = 64

- $Q, K, V$: (50, 64) each
- **$QK^T$**: (50, 64) x (64, 50) = **(50, 50)**, one score for every pair of tokens
- **Output** = softmax(...)(50, 50) x V(50, 64) = **(50, 64)**

(With a batch dimension B these become (B, 50, 50) and (B, 50, 64).)

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
L, d, dk = 50, 64, 64
x = tf.random.normal((1, L, d))          # batch of 1 sequence

Wq, Wk, Wv = layers.Dense(dk), layers.Dense(dk), layers.Dense(dk)
Q, K, V = Wq(x), Wk(x), Wv(x)
```

```python
# The 2-line attention snippet
weights = tf.nn.softmax(tf.matmul(Q, K, transpose_b=True) / tf.sqrt(float(dk)), axis=-1)
output = tf.matmul(weights, V)

print("Q K^T shape :", tf.matmul(Q, K, transpose_b=True).shape)
print("Output shape:", output.shape)
print("Each row of weights sums to 1:", np.allclose(tf.reduce_sum(weights, -1).numpy(), 1.0))
```
