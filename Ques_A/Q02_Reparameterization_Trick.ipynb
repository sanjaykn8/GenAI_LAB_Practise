# Question 2 - The Reparameterization Trick
### 1. Formula
$$z = \mu + \sigma \odot \epsilon, \qquad \sigma = \exp\left(\tfrac{1}{2}\log\sigma^2\right), \qquad \epsilon \sim \mathcal{N}(0, I)$$
The randomness is moved into $\epsilon$, so $z$ is a differentiable function of $\mu$ and $\log\sigma^2$.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
def reparameterize(mu, logvar):
    eps = tf.random.normal(shape=tf.shape(mu))
    return mu + tf.exp(0.5 * logvar) * eps
```

```python
# Test: gradients flow back to mu and logvar
mu = tf.Variable(tf.zeros((4, 2)))
logvar = tf.Variable(tf.zeros((4, 2)))

with tf.GradientTape() as tape:
    z = reparameterize(mu, logvar)
    loss = tf.reduce_sum(z ** 2)

grads = tape.gradient(loss, [mu, logvar])
print("z shape:", z.shape)
print("grad wrt mu is None?    ", grads[0] is None)
print("grad wrt logvar is None?", grads[1] is None)
```

```python
# Check the sample statistics: mu=3, sigma=2 (logvar = log 4)
z = reparameterize(tf.fill((100000, 1), 3.0), tf.fill((100000, 1), float(np.log(4.0))))
print("mean ~ 3:", float(tf.reduce_mean(z)), " std ~ 2:", float(tf.math.reduce_std(z)))
```
