# Question 1 - Two-player Minimax Game
Objective: $V(x,y) = x^2 - y^2 + xy$. Player A **minimizes** V by updating x, Player B **maximizes** V by updating y.

### 1. Gradients
- $\dfrac{\partial V}{\partial x} = 2x + y$
- $\dfrac{\partial V}{\partial y} = -2y + x$

Update rules (simultaneous gradient descent / ascent):
- $x \leftarrow x - \eta (2x + y)$  (A descends)
- $y \leftarrow y + \eta (x - 2y)$  (B ascends)

```python
import numpy as np
import matplotlib.pyplot as plt

x, y = 2.0, 2.0
lr = 0.1
history = [(x, y)]

for i in range(20):
    dV_dx = 2 * x + y
    dV_dy = -2 * y + x
    x = x - lr * dV_dx      # player A minimizes
    y = y + lr * dV_dy      # player B maximizes
    history.append((x, y))
    print(f"iter {i+1:2d}: x = {x:.5f}, y = {y:.5f}, V = {x**2 - y**2 + x*y:.5f}")
```

```python
h = np.array(history)
plt.figure(figsize=(5, 5))
plt.plot(h[:, 0], h[:, 1], 'o-')
plt.scatter([0], [0], c='red', marker='*', s=200, label='(0,0)')
plt.xlabel('x'); plt.ylabel('y'); plt.title('Trajectory of (x, y)')
plt.legend(); plt.grid(True); plt.show()
```

```python
# Why: the update is linear -> [x, y]_new = M [x, y]
M = np.array([[1 - 2*lr, -lr],
              [lr, 1 - 2*lr]])
eig = np.linalg.eigvals(M)
print("Update matrix M =\n", M)
print("Eigenvalues:", eig, " |lambda| =", np.abs(eig))
```

### 3. Converge or oscillate?
The update matrix is $M = \begin{bmatrix}0.8 & -0.1\\ 0.1 & 0.8\end{bmatrix}$ with eigenvalues $0.8 \pm 0.1i$, $|\lambda| \approx 0.806 < 1$.

So **(x, y) converges to (0, 0)**. The complex eigenvalues add a small rotation (a slight spiral), but there is no lasting oscillation, because V is strongly convex in x ($+x^2$) and strongly concave in y ($-y^2$). After 20 steps both values are close to 0 (about 0.04 or less).
