# Question 12 - DCGAN Shape Tracking (Fashion-MNIST)
$H_{out} = (H_{in}-1)s - 2p + k$

- Layer 1: Dense 100 -> 6272, reshaped to **(128, 7, 7)**
- **Layer 2**: $(7-1)2 - 2 + 4 = 14$, giving **(C, H, W) = (64, 14, 14)**
- **Layer 3**: $(14-1)2 - 2 + 4 = 28$, giving **(C, H, W) = (1, 28, 28)**

Keras uses channels-last, so it prints (14, 14, 64) and (28, 28, 1).

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
class DCGANGenerator(keras.Model):
    def __init__(self):
        super().__init__()
        self.l1 = layers.Dense(7 * 7 * 128, activation='relu')
        self.reshape = layers.Reshape((7, 7, 128))
        self.l2 = layers.Conv2DTranspose(64, 4, strides=2, padding='same', activation='relu')  # = PyTorch padding=1
        self.l3 = layers.Conv2DTranspose(1, 4, strides=2, padding='same', activation='tanh')

    def call(self, z):
        x = self.reshape(self.l1(z)); print("Layer 1:", x.shape)
        x = self.l2(x);               print("Layer 2:", x.shape)
        x = self.l3(x);               print("Layer 3:", x.shape)
        return x

img = DCGANGenerator()(tf.random.normal((1, 100)))
```

```python
# Normalize Fashion-MNIST to [-1, 1] for a Tanh output
(x_train, _), _ = keras.datasets.fashion_mnist.load_data()
x_train = x_train.astype("float32") / 127.5 - 1.0
print("range:", x_train.min(), x_train.max())
```

### 2. Normalization range
Tanh outputs values in **[-1, 1]**, so Fashion-MNIST images must be normalized to **[-1, 1]** with `x / 127.5 - 1`.
